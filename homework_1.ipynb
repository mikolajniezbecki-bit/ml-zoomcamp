```python
median_before = df['horsepower'].median()
print("Mediana przed:", median_before)
```

```python
most_frequent = df['horsepower'].mode()[0]
print("Najczęstsza wartość:", most_frequent)
```

```python
median_after = df['horsepower'].fillna(most_frequent).median()
print("Mediana po:", median_after)
```

```python
if median_after > median_before:
    print("Wynik: Yes, it increased")
elif median_after < median_before:
    print("Wynik: Yes, it decreased")
else:
    print("Wynik: No")
```

```python
asia_cars = df[df['origin'] == 'Asia']
```

```python
selected = asia_cars[['vehicle_weight', 'model_year']].iloc[:7]
```

```python
X = selected.values
print(X)
```

```python
XTX = X.T.dot(X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
```

```python
result = w.sum()
print("Suma wag:", result)
```
